```python
def temizlik_hatti(df):
    try:
        # Sütun isimlerini standartlaştır
        df.columns = df.columns.str.strip().str.replace(' ', '_').str.upper()
        
        # Artık 'MINIMUM_HIZ' yerine 'MINIMUM_SPEED' kullanmalısın
        ortalama_hiz = df['MINIMUM_SPEED'].mean()
        df['MINIMUM_SPEED'] = df['MINIMUM_SPEED'].fillna(ortalama_hiz)
        
        # Hatalı hızları düzelt
        df['MINIMUM_SPEED'] = df['MINIMUM_SPEED'].apply(lambda x: 0 if x < 0 else x)
        
        print("Temizlik başarıyla tamamlandı.")
        return df
    except Exception as e:
        print(f"Hata oluştu: {e}")
        return None
```
