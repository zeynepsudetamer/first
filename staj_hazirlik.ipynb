```python
# 1. Gün: Python'a Giriş
# Listeler ve Sözlükler (Veri bilimi için en temel yapılar)

# Örnek: Akıllı şehir trafik verisi (temsili)
saatler = [8, 9, 10, 11, 12]
arac_sayilari = [150, 200, 1350, 300, 443]

# İki listeyi birleştirip sözlük (dictionary) haline getirelim
trafik_durumu = dict(zip(saatler, arac_sayilari))

# Döngü ile veriyi ekrana yazdıralım
for saat, sayi in trafik_durumu.items():
    if sayi>300:
      print(f"Saat {saat}:00'de trafik yoğunluğu: {sayi} araç.")
    
```
