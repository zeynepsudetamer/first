```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Veriyi oku
df = pd.read_csv('traffic_density_202307.csv')

# 2. Tarih sütununu çevir ve saati al (Doğru sütun adı: DATE_TIME)
df['DATE_TIME'] = pd.to_datetime(df['DATE_TIME'])
df['Saat'] = df['DATE_TIME'].dt.hour

# 3. Gruplama yap (Doğru sütun adı: NUMBER_OF_VEHICLES)
saatlik_analiz = df.groupby('Saat')['NUMBER_OF_VEHICLES'].mean()

# 4. Grafiği çizdir
plt.figure(figsize=(10, 6))
saatlik_analiz.plot(kind='line', marker='o', color='b')
plt.title('Gunluk Ortalama Arac Yogunlugu')
plt.xlabel('Saat')
plt.ylabel('Arac Sayisi')
plt.grid(True)
plt.show()
```

```python
hiz_analizi=df.groupby('Saat')['AVERAGE_SPEED'].mean()
plt.figure(figsize=(10,6))
hiz_analizi.plot(kind='line',marker='o',color='r',linestyle='--')
plt.title('Gunluk Ortalama Hiz Degisimi')
plt.xlabel('Saat')
plt.ylabel('Ortalama Hiz (km/s')
plt.grid(True)
plt.show()
```
